### Missing Value Imputation

- Median imputation is used for the missing **numerical values.**
- Median is preferred because it is less affected by **outliers** and **extreme values.**
- This helps retain the existing data without removing observations.

```python
# Fill missing values using median
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())

# Verify missing values
df.isnull().sum()
```

### Initial EDA ***Checking Data Inconsistency-(Outliers)***

```python
### Outlier count using IQR method
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = ((df[col] < Q1 - 1.5 * IQR) | (df[col] > Q3 + 1.5 * IQR)).sum()
    print(f"{col}: {outliers} outliers")
```

### Outlier Detection

- Outliers were identified using the IQR method.
- Some numerical variables contain outliers, while others have none.
- Higher outlier counts are observed in ***Yield, Production, Revenue, Profit, Water Used, and Water Efficiency.***
- These values will be further examined to determine whether they represent genuine observations or data errors.

### Based on EDA Final Decision for Outlier Handling

- Outliers were identified using the IQR (Interquartile Range) method across the numerical variables.
- Several outliers were observed in variables such as Yield, Production, Revenue, Profit, Water Usage, and Water Efficiency.
- These extreme values were further inspected using their minimum and maximum values.
- The observed extreme values are considered **potentially valid agricultural observations** rather than obvious data-entry errors. For example, higher values of Yield, Production, Revenue, and Profit can occur due to differences in crop type, farm size, agricultural practices, and productivity.
- Similarly, negative Profit values can represent genuine losses and should not be treated as incorrect observations.
- Therefore, the detected outliers were **not removed or capped**, as doing so could lead to loss of meaningful information and distort the actual agricultural patterns in the dataset.
- The original values are retained for further analysis to preserve the natural variation present in the agricultural data.

### Feature Engineering

- **Feature engineering is performed to create meaningful derived variables from the existing dataset.**
- **These new features help provide a better understanding of agricultural profitability, production cost, and farm-level performance.**
- **Profit Margin is created to measure the percentage of revenue retained as profit.**
- **Cost per Tonne is created to measure the production cost associated with each tonne of agricultural output.**
- **Revenue per Hectare is created to compare revenue generated across farms of different sizes.**
- **Profit per Hectare is created to compare farm-level profitability independent of farm size.**
- **These features are derived from existing variables and do not introduce external information into the dataset.**

## **Create the features**

```python
# Create derived features for profitability and farm-level performance

df['Profit_Margin_pct'] = (df['Profit_INR'] / df['Revenue_INR']) * 100

df['Cost_per_Tonne'] = df['Total_Cost_INR'] / df['Production_Tonnes']

df['Revenue_per_Hectare'] = df['Revenue_INR'] / df['Farm_Area_Hectares']

df['Profit_per_Hectare'] = df['Profit_INR'] / df['Farm_Area_Hectares']

df.head()
```

```python
df[
    [
        'Profit_Margin_pct',
        'Cost_per_Tonne',
        'Revenue_per_Hectare',
        'Profit_per_Hectare'
    ]
].describe().round(2)
```

```python
# Check for missing and infinite values in engineered features

engineered_cols = [
    'Profit_Margin_pct',
    'Cost_per_Tonne',
    'Revenue_per_Hectare',
    'Profit_per_Hectare'
]

print("Missing values:")
print(df[engineered_cols].isnull().sum())

print("\nInfinite values:")
print(df[engineered_cols].isin([float('inf'), float('-inf')]).sum())
```

### Feature Engineering Validation

- **All newly created features contain 0 missing values and 0 infinite values.**
- **This confirms that the feature calculations were performed correctly without generating invalid values.**

### ***Statistical Significance Analysis***

- **Statistical testing is performed to determine whether the observed differences in agricultural performance across seasons are statistically significant.**
- **One-Way ANOVA is used to compare the mean Yield and Profit across Kharif, Rabi, and Zaid seasons.**
- **The test helps determine whether the differences observed in the season-wise analysis are likely to represent meaningful differences rather than random variation in the sample.**
- **A significance level of 5% (α = 0.05) is used for hypothesis testing.**
- **If the p-value is less than 0.05, the null hypothesis is rejected, indicating a statistically significant difference among the seasonal means.**

### ANOVA for Yield

```python
from scipy.stats import f_oneway

yield_kharif = df[df['Season'] == 'Kharif']['Yield_Tonnes_Ha']
yield_rabi = df[df['Season'] == 'Rabi']['Yield_Tonnes_Ha']
yield_zaid = df[df['Season'] == 'Zaid']['Yield_Tonnes_Ha']

yield_stat, yield_p = f_oneway(
    yield_kharif,
    yield_rabi,
    yield_zaid
)

print("Yield ANOVA Statistic:", round(yield_stat, 4))
print("Yield ANOVA p-value:", round(yield_p, 6))
```

### ANOVA for Profit

```python
profit_kharif = df[df['Season'] == 'Kharif']['Profit_INR']
profit_rabi = df[df['Season'] == 'Rabi']['Profit_INR']
profit_zaid = df[df['Season'] == 'Zaid']['Profit_INR']

profit_stat, profit_p = f_oneway(
    profit_kharif,
    profit_rabi,
    profit_zaid
)

print("Profit ANOVA Statistic:", round(profit_stat, 4))
print("Profit ANOVA p-value:", round(profit_p, 6))
```

### Statistical Significance Analysis — Observations

- **One-Way ANOVA was performed to determine whether the differences in average Yield and Profit across Kharif, Rabi, and Zaid seasons are statistically significant.**
- **For Yield, the ANOVA produced an F-statistic of 1.5439 and a p-value of 0.213678. Since the p-value is greater than the 0.05 significance level, the null hypothesis is not rejected. Therefore, the observed differences in average Yield across seasons are not statistically significant based on this test.**
- **For Profit, the ANOVA produced an F-statistic of 34.2918 and an extremely small p-value (displayed as 0.0). Since the p-value is less than 0.05, the null hypothesis is rejected. Therefore, there is a statistically significant difference in average Profit across the three seasons.**
- **The results indicate that seasonal differences are statistically significant for Profit but not for Yield.**
- **This suggests that seasonal variation has a stronger statistical association with agricultural profitability than with average Yield in this dataset.**
- **ANOVA indicates that at least one seasonal mean differs significantly for Profit, but it does not identify which specific seasons differ from each other.**
